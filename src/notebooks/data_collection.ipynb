# Run simulation and get agent and model data

```python
# Cell 2: Run Simulation and Get Data
def run_evacuation_simulation():
    """Run evacuation simulation and return dataframes."""
    print("Creating model...")
    model = EvacuationModel(n_agents=40, width=30, height=25, num_exits=3, num_leaders=0,  dt=0.1)
    
    print(f"Starting simulation with {len(model.agents)} agents...")
    
    # Run simulation
    max_steps = 2000
    for i in range(max_steps):
        if i % 50 == 0:
            print(f"Step {i}: {len(model.agents)} agents remaining")
        
        model.step()
        
        if not model.running:
            print(f"Simulation completed at step {i}")
            break
    
    # Get dataframes
    model_data = model.datacollector.get_model_vars_dataframe()
    agent_data = model.datacollector.get_agent_vars_dataframe()
    
    print(f"\nSimulation Summary:")
    print(f"Total steps: {len(model_data)}")
    print(f"Agents evacuated: {model.n_agents - len(model.agents)}/{model.n_agents}")
    
    return model_data, agent_data, model

# Run the simulation
model_df, agent_df, final_model = run_evacuation_simulation()

# Display shape
print(f"\nModel DataFrame shape: {model_df.shape}")
print(f"Agent DataFrame shape: {agent_df.shape}")
```

# Analysis
## Basic stats

```python
print("=== EVACUATION STATISTICS ===\n")

# --- Simulation parameters ---
print(f"Simulation Duration: {len(model_df)} steps")
print(f"Time Step: {final_model.dt:.2f} seconds")
print(f"Total Time: {len(model_df) * final_model.dt:.1f} seconds")

# --- Agents statistics ---
print(f"\nAgent Statistics:")
print(f"Initial Agents: {model_df['Agents'].iloc[0]}")
print(f"Final Agents: {model_df['Agents'].iloc[-1]}")
print(f"Evacuation Rate: {(model_df['Agents'].iloc[0] - model_df['Agents'].iloc[-1]) / model_df['Agents'].iloc[0] * 100:.1f}%")

# --- Speed statistics ---
print(f"\nSpeed Statistics:")
print(f"Average Speed: {model_df['Average_Speed'].mean():.3f} m/s")
print(f"Max Speed: {model_df['Average_Speed'].max():.3f} m/s")
print(f"Min Speed: {model_df['Average_Speed'].min():.3f} m/s")

# --- Exit throughput and evacuation time ---
if "Exit_Flow_Total" in model_df.columns:
    total_evacuated = model_df["Exit_Flow_Total"].sum()
    print(f"\nThroughput:")
    print(f"Total Evacuated: {total_evacuated}")
    print(f"Mean Throughput per step: {model_df['Exit_Flow_Total'].mean():.2f} agents/step")

    # time when last agent evacuated (if all managed to leave)
    if total_evacuated == final_model.n_agents:
        evac_time_step = model_df.index[model_df["Exit_Flow_Total"].cumsum() >= final_model.n_agents][0]
        evac_time = evac_time_step * final_model.dt
        print(f"Total Evacuation Time: {evac_time:.2f} seconds")

# --- Exit usage balance (entropy measure) ---
if "Exit_Balance_Entropy" in model_df.columns:
    print(f"\nExit Usage Balance:")
    print(f"Mean Exit Balance Entropy: {model_df['Exit_Balance_Entropy'].mean():.3f}")

# --- Pressure near exits (crowding) ---
for exit_idx in range(3):  # support up to 3 exits
    col = f"Exit{exit_idx}_Pressure"
    if col in model_df.columns:
        print(f"Mean Pressure near Exit {exit_idx}: {model_df[col].mean():.2f}")

# --- Leadership and followers ---
if "Followers_Latched" in model_df.columns and "Leaders_Remaining" in model_df.columns:
    print(f"\nLeadership Dynamics:")
    print(f"Final Followers Latched: {model_df['Followers_Latched'].iloc[-1]}")
    print(f"Leaders Remaining at End: {model_df['Leaders_Remaining'].iloc[-1]}")

# --- Injuries ---
if "Injured_Total" in model_df.columns:
    print(f"\nInjury Statistics:")
    print(f"Total Injured: {model_df['Injured_Total'].iloc[-1]}")
    print(f" - Fire:  {model_df['Injured_Fire'].iloc[-1]}")
    print(f" - Smoke: {model_df['Injured_Smoke'].iloc[-1]}")
    print(f" - Pressure: {model_df['Injured_Press'].iloc[-1]}")

# --- Diversity effects (variation in parameters) ---
if "Std_v0_init" in model_df.columns:
    print(f"\nDiversity Measures (std of agent parameters):")
    print(f"Initial Desired Speed: {model_df['Std_v0_init'].iloc[-1]:.3f}")
    print(f"Max Speed:             {model_df['Std_vmax'].iloc[-1]:.3f}")
    print(f"Body Radius:           {model_df['Std_radius'].iloc[-1]:.3f}")
    print(f"Mass:                  {model_df['Std_mass'].iloc[-1]:.3f}")
    print(f"Visibility Radius:     {model_df['Std_vis'].iloc[-1]:.3f}")
    print(f"Panic Baseline:        {model_df['Std_panic0'].iloc[-1]:.3f}")
```

## Stats over time

```python
import numpy as np
import matplotlib.pyplot as plt

# --- helpers ---
dt = getattr(final_model, "dt", 0.05)
t  = model_df.index.values * dt
cols = set(model_df.columns)

def cumsum_if(col):
    """Cumulative evacuated for a per-step throughput column."""
    return model_df[col].cumsum() if col in cols else None

def has(*names):
    return all(n in cols for n in names)

# --- 1) Evacuation progress (agents remaining) ---
plt.figure(figsize=(10,5))
plt.plot(t, model_df['Agents'], linewidth=2)
plt.title('Evacuation Progress (Agents Remaining)')
plt.xlabel('Time [s]')
plt.ylabel('Agents')
plt.grid(True, alpha=0.3)
plt.show()

# --- 2) Average speed over time ---
plt.figure(figsize=(10,5))
plt.plot(t, model_df['Average_Speed'], linewidth=2)
plt.title('Average Speed')
plt.xlabel('Time [s]')
plt.ylabel('Speed [m/s]')
plt.grid(True, alpha=0.3)
plt.show()

# --- 3) Cumulative evacuated: total and per-exit (if available) ---
plt.figure(figsize=(10,5))
if 'Exit_Flow_Total' in cols:
    cum_total = model_df['Exit_Flow_Total'].cumsum()
else:
    # fallback from Agents
    cum_total = (model_df['Agents'].iloc[0] - model_df['Agents'])
plt.plot(t, cum_total, linewidth=2, label='Total')

# per-exit cumulatives if present
for k in range(3):
    name = f'Exit{k}_Flow'
    if name in cols:
        plt.plot(t, model_df[name].cumsum(), linewidth=1.5, linestyle='--', label=f'Exit {k}')

plt.title('Cumulative Evacuated (Total & per Exit)')
plt.xlabel('Time [s]')
plt.ylabel('Evacuated agents')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# --- 4) Instantaneous throughput with smoothing (rolling mean) ---
window = max(1, int(2.0/dt))  # ~2s window
plt.figure(figsize=(10,5))
if 'Exit_Flow_Total' in cols:
    flow_s = model_df['Exit_Flow_Total'].rolling(window, min_periods=1).mean()
    plt.plot(t, flow_s, linewidth=2, label='Total (smoothed)')
for k in range(3):
    name = f'Exit{k}_Flow'
    if name in cols:
        plt.plot(t, model_df[name].rolling(window, min_periods=1).mean(),
                 linewidth=1.5, linestyle='--', label=f'Exit {k} (smoothed)')
plt.title('Exit Throughput (rolling mean)')
plt.xlabel('Time [s]')
plt.ylabel('Agents / step')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# --- 5) Pressure near exits ---
any_pressure = False
plt.figure(figsize=(10,5))
for k in range(3):
    name = f'Exit{k}_Pressure'
    if name in cols:
        any_pressure = True
        plt.plot(t, model_df[name], linewidth=1.8, label=f'Exit {k}')
if any_pressure:
    plt.title('Mean Radial Pressure near Exits')
    plt.xlabel('Time [s]')
    plt.ylabel('Pressure [N/m]')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
else:
    plt.close()

# --- 6) Leaders remaining & followers latched ---
plt.figure(figsize=(10,5))
if 'Leaders_Remaining' in cols:
    plt.plot(t, model_df['Leaders_Remaining'], linewidth=2, label='Leaders remaining')
if 'Followers_Latched' in cols:
    plt.plot(t, model_df['Followers_Latched'], linewidth=2, label='Followers latched')
plt.title('Leaders & Latching Dynamics')
plt.xlabel('Time [s]')
plt.ylabel('Count')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# --- 7) Visibility term & smoke exposure ---
plt.figure(figsize=(10,5))
if 'Mean_VisTerm' in cols:
    plt.plot(t, model_df['Mean_VisTerm'], linewidth=2, label='Mean visibility term')
if 'Mean_SmokeExposure' in cols:
    plt.plot(t, model_df['Mean_SmokeExposure'], linewidth=2, label='Mean smoke exposure')
plt.title('Environment Signals (Visibility & Smoke)')
plt.xlabel('Time [s]')
plt.ylabel('Value')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# --- 8) Injuries over time (totals & causes) ---
any_injury = False
plt.figure(figsize=(10,5))
for name, lab in [
    ('Injured_Total', 'Total injured'),
    ('Injured_Press', 'Pressure'),
    ('Injured_Fire',  'Fire'),
    ('Injured_Smoke', 'Smoke'),
]:
    if name in cols:
        any_injury = True
        plt.plot(t, model_df[name], linewidth=2, label=lab)
if any_injury:
    plt.title('Injury Counts Over Time')
    plt.xlabel('Time [s]')
    plt.ylabel('Count')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
else:
    plt.close()

# --- 9) Exit usage balance (entropy) ---
if 'Exit_Balance_Entropy' in cols:
    plt.figure(figsize=(10,5))
    plt.plot(t, model_df['Exit_Balance_Entropy'], linewidth=2)
    plt.title('Exit Usage Balance (Shannon entropy)')
    plt.xlabel('Time [s]')
    plt.ylabel('Entropy [nats]')
    plt.grid(True, alpha=0.3)
    plt.show()

# --- 10) Diversity (std) time series (if you keep them time-varying) ---
div_cols = ['Std_v0_init','Std_vmax','Std_radius','Std_mass','Std_vis','Std_panic0']
if any(c in cols for c in div_cols):
    plt.figure(figsize=(10,5))
    for c in div_cols:
        if c in cols:
            plt.plot(t, model_df[c], linewidth=1.8, label=c.replace('Std_','std '))
    plt.title('Population Diversity (std over time)')
    plt.xlabel('Time [s]')
    plt.ylabel('Standard deviation')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
```

## Trajectory analysis

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# --- read sizes/layout from the model ---
W  = float(getattr(final_model, "width", 20.0))
H  = float(getattr(final_model, "height", 15.0))
walls = list(getattr(final_model, "walls", []))
exits = list(getattr(final_model, "exits", []))

def draw_layout(ax):
    """Draw room walls and exits on given axes."""
    for (x0, y0, x1, y1) in walls:
        ax.plot([x0, x1], [y0, y1], 'k-', linewidth=3, alpha=0.9, zorder=3)
    for (x0, y0, x1, y1) in exits:
        ax.plot([x0, x1], [y0, y1], 'g-', linewidth=6, alpha=0.9, zorder=4)
    ax.set_xlim(-0.5, W + 0.5)
    ax.set_ylim(-0.5, H + 0.5)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.25)

def smooth_like_gaussian(img):
    """Tiny 3x3 Gaussian-ish blur (no SciPy). Keeps edges via reflect padding."""
    k = np.array([[1, 2, 1],
                  [2, 4, 2],
                  [1, 2, 1]], dtype=float)
    k /= k.sum()
    pad = 1
    p = np.pad(img, pad_width=pad, mode="reflect")
    out = (
        k[0,0]*p[:-2, :-2] + k[0,1]*p[:-2, 1:-1] + k[0,2]*p[:-2, 2:] +
        k[1,0]*p[1:-1, :-2] + k[1,1]*p[1:-1, 1:-1] + k[1,2]*p[1:-1, 2:] +
        k[2,0]*p[2:, :-2] + k[2,1]*p[2:, 1:-1] + k[2,2]*p[2:, 2:]
    )
    return out

# --- figure ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# =========================
# Panel 1: Sample trajectories
# =========================
if not agent_df.empty:
    # choose up to N agents to plot
    N = 20
    unique_agents = agent_df.index.get_level_values('AgentID').unique()[:N]

    # discover leaders using the first row per agent (DataCollector has 'is_leader')
    leader_flags = {}
    for aid in unique_agents:
        try:
            row0 = agent_df.xs(aid, level='AgentID').iloc[0]
            leader_flags[aid] = bool(row0.get('is_leader', False))
        except Exception:
            leader_flags[aid] = False

    colors = plt.cm.tab10(np.linspace(0, 1, len(unique_agents)))
    for i, agent_id in enumerate(unique_agents):
        traj = agent_df.xs(agent_id, level='AgentID')
        ax1.plot(
            traj['x'].to_numpy(), traj['y'].to_numpy(),
            linewidth=2, alpha=0.85, color=colors[i],
            label=f"Agent {agent_id}" + (" (L)" if leader_flags.get(agent_id, False) else "")
        )
        # start (o) and end (x)
        ax1.scatter(traj['x'].iloc[0],  traj['y'].iloc[0],
                    s=90, marker='o', alpha=0.95, color=colors[i], zorder=5)
        ax1.scatter(traj['x'].iloc[-1], traj['y'].iloc[-1],
                    s=90, marker='x', alpha=0.95, color=colors[i], zorder=5)

draw_layout(ax1)
ax1.set_title('Sample Agent Trajectories', fontsize=14)
if not agent_df.empty:
    ax1.legend(bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0.)
ax1.set_xlabel('X [m]'); ax1.set_ylabel('Y [m]')

# =========================
# Panel 2: Heatmap of positions (exclude injured frames)
# =========================
if not agent_df.empty:
    # Build an "active" subset: only samples BEFORE the first injury of each agent
    df = agent_df.reset_index()  # expects columns: Step, AgentID, x, y, Speed, injured, ...

    # First step where each agent becomes injured (if any)
    first_injury = df[df["injured"]].groupby("AgentID")["Step"].min()  # Series: AgentID -> step
    df["inj_step"] = df["AgentID"].map(first_injury)                   # NaN if never injured

    # Keep rows strictly before injury (or all rows if never injured)
    active = df[(df["Step"] < df["inj_step"]) | df["inj_step"].isna()]

    # Optional: ignore near-standstill samples to reduce “hot pixels”
    speed_eps = 0.05
    active = active[active["Speed"] >= speed_eps]

    # If you prefer to drop all agents that ever got injured, uncomment:
    # injured_agents = set(df.loc[df["injured"], "AgentID"].unique())
    # active = active[~active["AgentID"].isin(injured_agents)]

    # 2D histogram on active samples only
    all_x = active["x"].to_numpy()
    all_y = active["y"].to_numpy()

    bins_x = int(np.clip(W * 2.0, 40, 140))
    bins_y = int(np.clip(H * 2.0, 30, 140))

    hist, xedges, yedges = np.histogram2d(
        all_x, all_y,
        bins=[bins_x, bins_y],
        range=[[0, W], [0, H]]
    )

    # Slight Gaussian-like smoothing
    smooth = smooth_like_gaussian(hist)

    # Percentile clipping: prevents few high values from flattening the rest
    pos = smooth[smooth > 0]
    if pos.size:
        vmin = np.percentile(pos, 5)
        vmax = np.percentile(smooth, 99.5)
        if vmax <= vmin:
            vmax = vmin + 1e-6
    else:
        vmin, vmax = 0.0, 1.0

    norm = mcolors.PowerNorm(gamma=0.6, vmin=vmin, vmax=vmax)

    im = ax2.imshow(
        smooth.T, origin='lower', aspect='equal',
        extent=[0, W, 0, H],
        cmap='inferno', norm=norm, alpha=0.96, zorder=1
    )
    cb = plt.colorbar(im, ax=ax2, fraction=0.046, pad=0.04)
    cb.set_label('Agent visits (active only, smoothed, γ=0.6)')

draw_layout(ax2)
ax2.set_title('Agent Position Heatmap', fontsize=14)
ax2.set_xlabel('X [m]'); ax2.set_ylabel('Y [m]')
```
